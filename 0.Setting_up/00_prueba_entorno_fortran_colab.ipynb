## 2. Patrón que usaremos las Semanas 1–8: escribir → compilar → ejecutar

Vamos a escribir el código FORTRAN en un archivo `.f90` (con la magia `%%writefile`), compilarlo con `gfortran` y correr el ejecutable — el mismo ciclo que usarías en tu propio computador, solo que aquí las tres celdas quedan documentadas en el notebook.

**Nota:** la primera línea de una celda `%%writefile` debe ser exactamente eso — no puede llevar nada más en esa línea.

```python
%%writefile hola.f90
program hola
    implicit none
    print *, "Hola desde FORTRAN, corriendo en Colab"
end program hola
```

```python
!gfortran hola.f90 -o hola -O2
!./hola
```

## 3. Celda de verificación (PASS/FAIL)

Cada taller del curso debe terminar con 2–3 casos de prueba conocidos que impriman `PASS`/`FAIL`, para poder corregir sin rastrear la lógica a mano. Aquí un ejemplo mínimo: un programa FORTRAN que calcula una suma conocida (1+2+...+100 = 5050), escribe el resultado a un archivo, y una celda de Python que lo lee y verifica.

```python
%%writefile suma.f90
program suma
    implicit none
    integer :: i, total
    total = 0
    do i = 1, 100
        total = total + i
    end do
    open(unit=10, file="resultado.txt", status="replace")
    write(10, *) total
    close(10)
end program suma
```

```python
!gfortran suma.f90 -o suma -O2
!./suma
!cat resultado.txt
```

```python
# Verificación PASS/FAIL (se corre desde Python, leyendo la salida de FORTRAN)
with open("resultado.txt") as f:
    resultado = int(f.read().strip())

esperado = 5050
if resultado == esperado:
    print(f"PASS: suma 1..100 = {resultado}")
else:
    print(f"FAIL: se obtuvo {resultado}, se esperaba {esperado}")
```
